```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/bluestock_mf.db")

queries = [
    ("Avg NAV per fund", """
        SELECT amfi_code, ROUND(AVG(nav),2) as avg_nav
        FROM fact_nav GROUP BY amfi_code ORDER BY avg_nav DESC"""),
    
    ("Top 5 SIP investors by amount", """
        SELECT investor_id, SUM(amount_inr) as total
        FROM fact_transactions WHERE transaction_type='SIP'
        GROUP BY investor_id ORDER BY total DESC LIMIT 5"""),
    
    ("Transactions by state", """
        SELECT state, COUNT(*) as total
        FROM fact_transactions
        GROUP BY state ORDER BY total DESC"""),
    
    ("Funds with expense ratio < 1%", """
        SELECT scheme_name, expense_ratio_pct
        FROM dim_fund WHERE expense_ratio_pct < 1.0
        ORDER BY expense_ratio_pct"""),
    
    ("Monthly transaction volume", """
        SELECT strftime('%Y-%m', transaction_date) as month,
        COUNT(*) as total
        FROM fact_transactions
        GROUP BY month ORDER BY month"""),
    
    ("T30 vs B30 transaction split", """
        SELECT city_tier, COUNT(*) as count,
        ROUND(AVG(amount_inr),2) as avg_amount
        FROM fact_transactions GROUP BY city_tier""")
]

for title, q in queries:
    print(f"\n{'='*40}")
    print(f"Query: {title}")
    print(pd.read_sql(q, conn).to_string(index=False))
```

```python
queries = [
    ("Avg NAV per fund", """
        SELECT amfi_code, ROUND(AVG(nav),2) as avg_nav
        FROM fact_nav GROUP BY amfi_code ORDER BY avg_nav DESC"""),
    
    ("Top 5 SIP investors by amount", """
        SELECT investor_id, SUM(amount_inr) as total
        FROM fact_transactions WHERE transaction_type='SIP'
        GROUP BY investor_id ORDER BY total DESC LIMIT 5"""),
    
    ("Transactions by state", """
        SELECT state, COUNT(*) as total
        FROM fact_transactions
        GROUP BY state ORDER BY total DESC"""),
    
    ("Funds with expense ratio < 1%", """
        SELECT scheme_name, expense_ratio_pct
        FROM dim_fund WHERE expense_ratio_pct < 1.0
        ORDER BY expense_ratio_pct"""),
    
    ("Monthly transaction volume", """
        SELECT strftime('%Y-%m', transaction_date) as month,
        COUNT(*) as total
        FROM fact_transactions
        GROUP BY month ORDER BY month"""),
    
    ("T30 vs B30 transaction split", """
        SELECT city_tier, COUNT(*) as count,
        ROUND(AVG(amount_inr),2) as avg_amount
        FROM fact_transactions GROUP BY city_tier""")
]

for title, q in queries:
    print(f"\n{'='*40}")
    print(f"Query: {title}")
    print(pd.read_sql(q, conn).to_string(index=False))
```
